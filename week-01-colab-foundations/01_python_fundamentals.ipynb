## 2. Statements

A statement is an instruction that Python can execute. Assignment statements store values in variables. Function-call statements ask a function to perform an action.

```python
course_name = "AI Engineering"  # Assignment statement
lesson_number = 1
print(course_name, "- Lesson", lesson_number)  # Function call
```

### Continuing a statement across lines

Parentheses, brackets, and braces allow clean multiline expressions. A backslash also works, but brackets are usually easier to maintain. Python permits semicolon-separated statements, but one statement per line is clearer.

```python
total = (
    1 + 2 + 3
    + 4 + 5 + 6
    + 7 + 8 + 9
)

colors = [
    "red",
    "blue",
    "green",
]

print("Total:", total)
print("Colors:", colors)
```

## 3. Indentation

Python uses indentation to show which statements belong to a block. Use four spaces for each indentation level.

```python
for number in range(1, 6):
    square = number ** 2
    print(number, "squared is", square)

print("The loop is complete.")
```

The indented lines run during every loop iteration. The final `print()` is outside the loop because it is not indented.

## 4. Comments and docstrings

A Python comment begins with `#`. Triple-quoted text is a string, not a multiline comment. When it appears first inside a function, class, or module, it becomes a docstring that documents that object.

```python
# This comment explains why the variable exists.
welcome_text = "Hello from GOMYCODE"

def build_welcome_message(student_name):
    """Return a personalized welcome message."""
    return f"{welcome_text}, {student_name}!"

print(build_welcome_message("Samuel"))
print("Function documentation:", build_welcome_message.__doc__)
```

## 5. Imports

A module groups reusable Python definitions. Import the module when you want its namespace, or import a specific name when that makes the code clearer.

```python
import math
from math import pi

radius = 3
area = pi * radius ** 2

print("Pi from the module:", math.pi)
print("Circle area:", round(area, 2))
```
